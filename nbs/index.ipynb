## How to use

Turn an academic paper into a SolveIt notebook you can read incrementally, with its citations footnoted and its bibliography folded away.

Inside SolveIt, convert a paper into blocks in the dialog you're reading:

```python
from kmacsolves.paper2solveit import *

await paper2solveit("https://arxiv.org/abs/2504.20997")   # a public URL

list_zotero('Reading List')                               # or your own library
await zotero_pdf(3)
```

From a terminal, the same pipeline writes a notebook file instead:

```sh
$ zot-ls                      # which collections do I have?
$ zot-ls '0.1 Inbox'          # what's in this one, and what has a PDF?
$ zot-nb 3                    # -> 'Vaswani17 - Attention Is All You Need.ipynb'

$ paper-nb https://arxiv.org/abs/2504.20997
```

Conversion uses the [Datalab Marker](https://www.datalab.to) OCR API — put `DATALAB_KEY` in your environment (about 10¢ per paper). Reading your own library needs either the Zotero desktop app with **Settings → Advanced → Allow other applications on this computer to communicate with Zotero** switched on, or `ZOTERO_USER_ID` and `ZOTERO_API_KEY` from [zotero.org/settings/keys](https://www.zotero.org/settings/keys) for the web API.
